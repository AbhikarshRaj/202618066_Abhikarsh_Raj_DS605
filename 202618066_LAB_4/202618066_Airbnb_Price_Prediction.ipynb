```python
from sklearn.model_selection import train_test_split

# Features
X = df_model.drop(
    columns=["price", "log_price"],
    errors="ignore"
)

# Target
y = df_model["price"]

# 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# Text feature
text_feature = "name"

# Categorical features
categorical_features = [
    "neighbourhood_group",
    "neighbourhood",
    "room_type"
]

# Numeric features
numeric_features = [
    col for col in X_train.columns
    if col not in categorical_features + [text_feature]
]

print("Numeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)

print("\nText feature:")
print(text_feature)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)
```

```python
categorical_transformer = Pipeline(
    steps=[
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )
    ]
)
```

```python
text_transformer = TfidfVectorizer(
    max_features=3000,
    ngram_range=(1, 2),
    min_df=2
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),

        (
            "cat",
            categorical_transformer,
            categorical_features
        ),

        (
            "text",
            text_transformer,
            text_feature
        )
    ]
)
```

```python
X_train_encoded = preprocessor.fit_transform(X_train)

print("Original X_train shape:", X_train.shape)
print("Encoded X_train shape:", X_train_encoded.shape)
```

```python
feature_names = preprocessor.get_feature_names_out()

print("Total encoded features:", len(feature_names))
print("\nFirst 30 features:")

for feature in feature_names[:30]:
    print(feature)
```

```python
import numpy as np
import pandas as pd
from scipy import sparse

y_array = y_train.to_numpy(dtype=float)

n = X_train_encoded.shape[0]

if sparse.issparse(X_train_encoded):

    X_enc = X_train_encoded.tocsr()

    mean_x = np.asarray(
        X_enc.mean(axis=0)
    ).ravel()

    mean_x2 = np.asarray(
        X_enc.power(2).mean(axis=0)
    ).ravel()

    mean_xy = np.asarray(
        X_enc.T @ y_array / n
    ).ravel()

else:

    X_enc = np.asarray(X_train_encoded)

    mean_x = X_enc.mean(axis=0)

    mean_x2 = (
        X_enc ** 2
    ).mean(axis=0)

    mean_xy = (
        X_enc.T @ y_array
    ) / n


mean_y = y_array.mean()

std_x = np.sqrt(
    np.maximum(
        mean_x2 - mean_x**2,
        0
    )
)

std_y = y_array.std()

correlations = (
    mean_xy - mean_x * mean_y
) / (std_x * std_y)

correlations = np.nan_to_num(correlations)
```

```python
corr_df = pd.DataFrame({
    "feature": feature_names,
    "correlation_with_price": correlations
})

corr_df["absolute_correlation"] = (
    corr_df["correlation_with_price"].abs()
)

corr_df = corr_df.sort_values(
    "absolute_correlation",
    ascending=False
)

corr_df.head(25)
```
