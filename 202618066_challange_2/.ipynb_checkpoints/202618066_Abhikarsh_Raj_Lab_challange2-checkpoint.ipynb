```python
# Pure-Python Bayesian optimization.
# Uses existing history, SPACE and evaluate().
# No additional imports. At most 10 NEW Oracle calls.

BO_BUDGET = 10
NAMES = list(SPACE)

# ---------- Basic numerical functions ----------

def ln_positive(x):
    if x <= 0:
        raise ValueError("Log-loss requires positive losses.")

    # Reduce x to [1, 2), then use a convergent log series.
    power = 0
    while x >= 2.0:
        x /= 2.0
        power += 1
    while x < 1.0:
        x *= 2.0
        power -= 1

    z = (x - 1.0) / (x + 1.0)
    term = z
    total = 0.0

    for denominator in range(1, 62, 2):
        total += term / denominator
        term *= z * z

    return 2.0 * total + power * 0.6931471805599453


def forward_solve(L, b):
    result = []
    for i in range(len(b)):
        subtotal = sum(L[i][j] * result[j] for j in range(i))
        result.append((b[i] - subtotal) / L[i][i])
    return result


def backward_solve(L, b):
    n = len(b)
    result = [0.0] * n

    for i in range(n - 1, -1, -1):
        subtotal = sum(L[j][i] * result[j] for j in range(i + 1, n))
        result[i] = (b[i] - subtotal) / L[i][i]

    return result


def factorize(K):
    n = len(K)

    # Retry with larger diagonal jitter if necessary.
    for jitter in (1e-8, 1e-6, 1e-4, 1e-2):
        L = [[0.0] * n for _ in range(n)]
        valid = True

        for i in range(n):
            for j in range(i + 1):
                value = K[i][j]
                if i == j:
                    value += jitter

                value -= sum(L[i][k] * L[j][k] for k in range(j))

                if i == j:
                    if value <= 0:
                        valid = False
                        break
                    L[i][j] = value ** 0.5
                else:
                    L[i][j] = value / L[j][j]

            if not valid:
                break

        if valid:
            return L

    raise ValueError("Surrogate factorization failed. No query made.")


# ---------- Mixed numeric/categorical representation ----------

def config_key(params):
    return tuple(params[name] for name in NAMES)


def encode_config(params):
    vector = []

    for name in NAMES:
        values = SPACE[name]

        if name in ("eta0", "alpha"):
            low = ln_positive(min(values))
            high = ln_positive(max(values))
            vector.append(
                (ln_positive(params[name]) - low) / (high - low)
                if high > low else 0.0
            )
        else:
            vector.append(values.index(params[name]))

    return vector


def kernel(a, b, length):
    numeric_distance = 0.0
    category_differences = 0

    for i, name in enumerate(NAMES):
        if name in ("eta0", "alpha"):
            numeric_distance += (a[i] - b[i]) ** 2
        elif a[i] != b[i]:
            category_differences += 1

    return (
        2.718281828459045 ** (
            -numeric_distance / (2.0 * length * length)
        )
        * (0.5 ** category_differences)
    )


# Enumerating candidates locally costs ZERO Oracle calls.
pool = [{}]
for name in NAMES:
    expanded = []
    for partial in pool:
        for value in SPACE[name]:
            candidate = partial.copy()
            candidate[name] = value
            expanded.append(candidate)
    pool = expanded

encoded_pool = [encode_config(params) for params in pool]


# ---------- Gaussian Process fitting and prediction ----------

def fit_gp(X, y, length):
    K = [
        [kernel(a, b, length) for b in X]
        for a in X
    ]
    L = factorize(K)
    coefficients = backward_solve(L, forward_solve(L, y))
    return X, L, coefficients, length


def predict_gp(model, x):
    X, L, coefficients, length = model

    k = [kernel(x, observed, length) for observed in X]
    mean = sum(a * b for a, b in zip(k, coefficients))

    solved = forward_solve(L, k)
    variance = max(1.0 - sum(v * v for v in solved), 0.0)

    return mean, variance


def choose_next():
    # Deduplicate for surrogate fitting without changing call history.
    observed = {}
    for trial in history:
        loss = trial["loss"]
        if not (0 < loss < float("inf")):
            raise ValueError("History contains nonpositive/nonfinite loss.")
        observed[config_key(trial["params"])] = trial

    trials = list(observed.values())
    X = [encode_config(t["params"]) for t in trials]
    logged = [ln_positive(t["loss"]) for t in trials]

    center = sum(logged) / len(logged)
    scale = (
        sum((value - center) ** 2 for value in logged) / len(logged)
    ) ** 0.5

    if scale < 1e-12:
        scale = 1.0

    y = [(value - center) / scale for value in logged]

    # Two smoothness assumptions reduce reliance on one fixed choice.
    models = [
        fit_gp(X, y, length=0.25),
        fit_gp(X, y, length=0.70)
    ]

    selected = None
    lowest_acquisition = float("inf")

    for params, encoded in zip(pool, encoded_pool):
        if config_key(params) in observed:
            continue

        predictions = [predict_gp(model, encoded) for model in models]
        mean = sum(p[0] for p in predictions) / len(predictions)

        # Includes uncertainty inside models and disagreement between them.
        variance = sum(
            p[1] + (p[0] - mean) ** 2
            for p in predictions
        ) / len(predictions)

        # Lower confidence bound: lower is more promising.
        acquisition = mean - 1.5 * variance ** 0.5

        if acquisition < lowest_acquisition:
            lowest_acquisition = acquisition
            selected = params.copy()

    return selected


# ---------- Sequential, budgeted Oracle evaluations ----------

bo_start_count = len(history)
bo_start_best = min(t["loss"] for t in history)

for iteration in range(BO_BUDGET):
    candidate = choose_next()

    if candidate is None:
        print("All configurations already tested.")
        break

    previous_best = min(t["loss"] for t in history)

    print("\nBayesian trial:", iteration + 1, "/", BO_BUDGET)
    print("Parameters:", candidate)

    # Uses your existing caching and JSON saving.
    # No automatic retry if a request fails.
    new_loss = evaluate(candidate)

    print("Improvement over best:", previous_best - new_loss)
    print("Best loss:", min(t["loss"] for t in history))

best = min(history, key=lambda t: t["loss"])

print("\n--- RESULTS ---")
print("Best parameters:", best["params"])
print("Best loss:", best["loss"])
print("Improvement during batch:", bo_start_best - best["loss"])
print("New recorded evaluations:", len(history) - bo_start_count)
print("Total recorded evaluations:", len(history))
```

```python
# Local neighborhood polish: one-swap-at-a-time until convergence.
# Reuses your existing `history`, `evaluate`, and `SPACE`.

MAX_POLISH_CALLS = 15
polish_start = len(history)

def already_tested(params):
    return any(t["params"] == params for t in history)

improved = True
round_num = 0
while improved and (len(history) - polish_start) < MAX_POLISH_CALLS:
    improved = False
    round_num += 1
    anchor = min(history, key=lambda t: t["loss"])["params"].copy()
    anchor_loss = min(t["loss"] for t in history)
    print(f"\n--- Round {round_num} | anchor loss = {anchor_loss:.8f} ---")

    for name in SPACE:
        for value in SPACE[name]:
            if value == anchor[name]:
                continue
            candidate = anchor.copy()
            candidate[name] = value
            if already_tested(candidate):
                continue
            if (len(history) - polish_start) >= MAX_POLISH_CALLS:
                break
            loss = evaluate(candidate)
            if loss < anchor_loss:
                print(f"  Improvement! {name}={value} -> {loss:.8f}")
                improved = True
                # keep scanning this round with the OLD anchor;
                # we'll re-anchor at the top of the next round
    if not improved:
        print("Converged: no single-parameter change improves the loss further.")

best = min(history, key=lambda t: t["loss"])
print("\n--- FINAL ---")
print("Best parameters:", best["params"])
print("Best loss:", best["loss"])
print("Total calls used for polish:", len(history) - polish_start)
print("Total recorded evaluations:", len(history))
```

## Summary

**Model:** SGDRegressor | **Team:** TEAM_65 | **Search space size:** 2,304 configurations

### Methodology

Optimization was performed in four stages, combining manual and automated search strategies
(no external HPO libraries used):

1. **Manual coordinate descent** — starting from a baseline configuration, individual
   hyperparameters were varied one at a time to quickly map out which regions of the space were
   viable vs. catastrophic (e.g. `loss='huber'` and `eta0=0.0001` in isolation caused severe
   loss blow-ups).
2. **One-swap local polish** — a systematic loop that tried every untested single-parameter
   change from the current best configuration, moving to any improvement found, repeated until a
   full round produced no further gains (confirmed convergence to a local optimum).
3. **Bayesian Optimization** — a from-scratch Gaussian Process surrogate (custom Cholesky-based
   GP regression over a mixed numeric/categorical kernel) with a lower-confidence-bound
   acquisition function, run in several budgeted batches to explore the space more broadly and
   escape local optima found by the polish step.
4. **Targeted single-parameter checks** — small, cheap follow-up tests (e.g. isolating
   `learning_rate`) to confirm no easy remaining gains were being missed.

### Notes on the search

- Improvement per call decreased sharply across later stages (from ~0.0035 loss reduction/call
  in early rounds to ~0.0002/call by the final Bayesian batch), indicating the search had
  converged to a plateau region near the optimum.
- The loss surface showed strong parameter interactions — several parameters that performed
  poorly in isolation (e.g. `loss='epsilon_insensitive'`, `eta0=0.0001`) turned out to be part of
  the best-performing combination once paired correctly with other hyperparameters, which a
  purely greedy one-at-a-time search could not have discovered alone.
- Search was stopped once successive optimization rounds showed consistently diminishing returns
  relative to the number of Oracle calls spent, balancing solution quality against the
  assignment's call-efficiency objective.
