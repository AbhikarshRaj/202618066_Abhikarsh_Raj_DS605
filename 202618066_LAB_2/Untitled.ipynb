**Task 2 - Vectorized Arithmetic and Linear Algebra**

**● Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @ or np.matmul().**

```python
import numpy as np
M1 = np.array([[1, 2, 3],
               [4, 5, 6]])
 
M2 = np.array([[7, 8, 9],
               [10, 11, 12]])
 
print("Matrix M1:\n", M1)
print("\nMatrix M2:\n", M2)
```

```python
addition = M1 + M2
print("\n--- M1 + M2 (Addition) ---")
print(addition)
 
```

```python
elementwise_mult = M1 * M2
print("\n--- M1 * M2 (Element-wise multiplication) ---")
print(elementwise_mult)
 
```

```python
M2_T = M2.T
print("\nM2 transposed (shape {}):\n{}".format(M2_T.shape, M2_T))
 
matmul_result_at = M1 @ M2_T
matmul_result_fn = np.matmul(M1, M2_T)
 
print("\n--- M1 @ M2.T (Matrix multiplication using @) ---")
print(matmul_result_at)
 
print("\n--- np.matmul(M1, M2.T) (same result, using function) ---")
print(matmul_result_fn)
 
print("\nResults match:", np.array_equal(matmul_result_at, matmul_result_fn))
```

**For a square matrix, find its transpose and determinant.**

```python
S = np.array([[2, 3, 1],
              [4, 1, 5],
              [6, 2, 8]])
 
print("Square Matrix S:\n", S)
print("Shape:", S.shape)
```

```python
S_transpose = S.T   # equivalent to np.transpose(S)
print("\n--- Transpose of S ---")
print(S_transpose)
 
```

```python
S_det = np.linalg.det(S)
print("\n--- Determinant of S ---")
print(S_det)
 
```

**If invertible, find the inverse and verify A @ A^-1 using np.allclose().**

```python
S = np.array([[2, 3, 1],
              [4, 1, 5],
              [6, 2, 8]])
 
print("Square Matrix S:\n", S)
S_det = np.linalg.det(S)
print("\nDeterminant:", S_det, "| Rounded:", round(S_det, 2))
 
# Inverse (only exists if determinant != 0)
if not np.isclose(S_det, 0):
    S_inv = np.linalg.inv(S)
    print("\n--- Inverse of S ---")
    print(S_inv)
 
    # Verify: S @ S_inv should equal the identity matrix
    identity_check = S @ S_inv
    print("\n--- S @ S_inv (should be identity) ---")
    print(identity_check)
 
    is_identity = np.allclose(identity_check, np.eye(S.shape[0]))
    print("\nS @ S_inv matches identity matrix (np.allclose):", is_identity)
else:
    print("\nDeterminant is 0 -> matrix is singular, no inverse exists.")
```

**Task 3 - Normal Distribution and Histogram /n**

**● Generate at least 1,000 values from a normal distribution. State the chosen mean and standard deviation.**

```python
import numpy as np
import matplotlib.pyplot as plt
 
# -----------------------------------------------------------
# Generate values from a normal distribution
# -----------------------------------------------------------
np.random.seed(42)          # reproducibility
 
chosen_mean = 50             # population mean (mu) we choose
chosen_std = 10               # population standard deviation (sigma) we choose
n_samples = 1000              # at least 1000 values
 
data = np.random.normal(loc=chosen_mean, scale=chosen_std, size=n_samples)
 
print("Chosen mean (mu):", chosen_mean)
print("Chosen std dev (sigma):", chosen_std)
print("Number of samples generated:", data.shape[0])
 
```

**● Find the sample mean and sample standard deviation. Compare them with the chosen values.**


```python
sample_mean = np.mean(data)
sample_std = np.std(data)
 
print("\n--- Sample Statistics ---")
print("Sample mean:", sample_mean)
print("Sample std dev:", sample_std)
 
print("\n--- Comparison with chosen values ---")
print("Mean difference   :", abs(sample_mean - chosen_mean))
print("Std dev difference:", abs(sample_std - chosen_std))
```


**● Plot a histogram with it.**

```python
plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, color="skyblue", edgecolor="black")
plt.axvline(sample_mean, color="red", linestyle="--", linewidth=2, label=f"Sample mean = {sample_mean:.2f}")
plt.title(f"Histogram of {n_samples} values from Normal Distribution (mu={chosen_mean}, sigma={chosen_std})")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.savefig("normal_distribution_histogram.png", dpi=150)
plt.show()
 
```

**● Male passengers older than 50: how many?**

```python
males_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]
print("--- Male passengers older than 50 ---")
print("Count:", males_over_50.shape[0])
 
```

```python
female_first_class = df[(df["Sex"] == "female") & (df["Pclass"] == 1)]
# Equivalent: df.query("Sex == 'female' and Pclass == 1")
 
female_first_class_count = female_first_class.shape[0]
female_first_class_survival_pct = female_first_class["Survived"].mean() * 100
 
print("\n--- Female first-class passengers ---")
print("Count:", female_first_class_count)
print(f"Survived: {female_first_class_survival_pct:.2f}%")
```

```python
fare_median = df["Fare"].median()
 
age_fare_survived = df[
    (df["Age"] >= 20) & (df["Age"] <= 40) &
    (df["Fare"] > fare_median) &
    (df["Survived"] == 1)
]
 
print("\n--- Age 20-40, Fare above overall median, survived ---")
print("Overall median fare:", fare_median)
print("Count:", age_fare_survived.shape[0])
```

```python
alone_young_not_survived = df[
    (df["SibSp"] == 0) & (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]
# Equivalent: df.query("SibSp == 0 and Parch == 0 and Age < 30 and Survived == 0")
 
print("\n--- Traveling alone, age < 30, did not survive ---")
print("Count:", alone_young_not_survived.shape[0])
```

```python
southampton_median_fare = df.loc[df["Embarked"] == "S", "Fare"].median()
 
southampton_filtered = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]
# Equivalent: df.query("Embarked == 'S' and Pclass in [2, 3] and Fare > @southampton_median_fare")
 
print("\n--- Embarked=S, Pclass 2 or 3, Fare above Southampton median ---")
print("Southampton median fare:", southampton_median_fare)
print("Count:", southampton_filtered.shape[0])
```

```python
survival_by_sex = df.groupby("Sex")["Survived"].mean()
print("--- Survival rate by Sex ---")
print(survival_by_sex)
 
```

```python
survival_by_pclass = df.groupby("Pclass")["Survived"].mean()
print("\n--- Survival rate by Pclass ---")
print(survival_by_pclass)
 
```

```python
age_fare_by_pclass = df.groupby("Pclass")[["Age", "Fare"]].mean()
print("\n--- Average Age and Fare by Pclass ---")
print(age_fare_by_pclass)
```

```python
sex_pclass_stats = df.groupby(["Sex", "Pclass"])["Survived"].agg(
    passenger_count="count",
    survival_rate="mean"
)
print("\n--- Passenger count and survival rate by Sex-Pclass ---")
print(sex_pclass_stats)
```

```python
age_missing_before = df["Age"].isnull().sum()
 
df_mean_filled = df.copy()
df_mean_filled["Age"] = df_mean_filled["Age"].fillna(df_mean_filled["Age"].mean())
 
age_missing_after = df_mean_filled["Age"].isnull().sum()
 
print("\n--- Age missing values: before vs after mean fill ---")
print("Before:", age_missing_before)
print("After :", age_missing_after)
 
```

```python

age_mean = df["Age"].mean()
age_median = df["Age"].median()
age_mode = df["Age"].mode()[0]   # mode() can return multiple values, take the first
 
np.random.seed(42)
non_null_ages = df["Age"].dropna().values
age_random_fill = df["Age"].apply(
    lambda x: np.random.choice(non_null_ages) if pd.isnull(x) else x
)
 
age_filled_mean = df["Age"].fillna(age_mean)
age_filled_median = df["Age"].fillna(age_median)
age_filled_mode = df["Age"].fillna(age_mode)
age_filled_random = age_random_fill
 
print("\n--- Age imputation comparison ---")
print(f"Mean used for fill  : {age_mean:.2f}")
print(f"Median used for fill: {age_median:.2f}")
print(f"Mode used for fill  : {age_mode:.2f}")
print("Random fill draws random existing Age values to fill NaNs")
```

```python
print("\nMissing after each method:")
print("Mean fill  :", age_filled_mean.isnull().sum())
print("Median fill:", age_filled_median.isnull().sum())
print("Mode fill  :", age_filled_mode.isnull().sum())
print("Random fill:", age_filled_random.isnull().sum())
 
print("\nResulting mean Age after each method:")
print("Mean fill  :", round(age_filled_mean.mean(), 2))
print("Median fill:", round(age_filled_median.mean(), 2))
print("Mode fill  :", round(age_filled_mode.mean(), 2))
print("Random fill:", round(age_filled_random.mean(), 2))
```

```python
# -----------------------------------------------------------
# Fare: Q1, Q3, IQR, 1.5*IQR bounds, and outlier count
# -----------------------------------------------------------
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1
 
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
 
fare_outliers = df[(df["Fare"] < lower_bound) | (df["Fare"] > upper_bound)]
 
print("\n--- Fare Outlier Analysis (IQR method) ---")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Outlier count:", fare_outliers.shape[0])
```

#Task 8 - Features and Pivot Table
● Create FamilySize = SibSp + Parch + 1 and IsAlone = 1 when FamilySize = 1.
● Create a pivot table: rows=Sex, columns=Pclass, values=mean Survived. Find the highest and lowest groups.

```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)
 
print("--- FamilySize and IsAlone (first 10 rows) ---")
print(df[["SibSp", "Parch", "FamilySize", "IsAlone"]].head(10))
 
print("\nIsAlone value counts:")
print(df["IsAlone"].value_counts())
```

```python
# -----------------------------------------------------------
# Pivot table: rows=Sex, columns=Pclass, values=mean Survived
# -----------------------------------------------------------
pivot = df.pivot_table(index="Sex", columns="Pclass", values="Survived", aggfunc="mean")
 
print("\n--- Pivot table: Survival rate by Sex and Pclass ---")
print(pivot)
```

```python
pivot_stacked = pivot.stack()   # flatten to a Series indexed by (Sex, Pclass)
 
highest_group = pivot_stacked.idxmax()
highest_value = pivot_stacked.max()
 
lowest_group = pivot_stacked.idxmin()
lowest_value = pivot_stacked.min()
 
print(f"\nHighest survival group: Sex={highest_group[0]}, Pclass={highest_group[1]} -> {highest_value:.4f}")
print(f"Lowest survival group : Sex={lowest_group[0]}, Pclass={lowest_group[1]} -> {lowest_value:.4f}")
 
```

```python
# -----------------------------------------------------------
# Correlation heatmap for relevant numerical columns
# -----------------------------------------------------------
numeric_cols = ["Survived", "Pclass", "Age", "SibSp", "Parch", "Fare"]
corr_matrix = df[numeric_cols].corr()
 
print("--- Correlation matrix ---")
print(corr_matrix)
 
fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr_matrix, cmap="coolwarm", vmin=-1, vmax=1)
 
ax.set_xticks(range(len(numeric_cols)))
ax.set_yticks(range(len(numeric_cols)))
ax.set_xticklabels(numeric_cols, rotation=45, ha="right")
ax.set_yticklabels(numeric_cols)
 
# annotate each cell with its correlation value
for i in range(len(numeric_cols)):
    for j in range(len(numeric_cols)):
        ax.text(j, i, f"{corr_matrix.iloc[i, j]:.2f}",
                 ha="center", va="center", color="black", fontsize=9)
 
ax.set_title("Correlation Heatmap - Numerical Columns")
fig.colorbar(im, ax=ax, label="Correlation coefficient")
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=150)
plt.show()
```

```python
# find strongest positive and negative relationships (excluding self-correlation of 1.0)
corr_values = np.array(corr_matrix.values, dtype=float, copy=True)
np.fill_diagonal(corr_values, np.nan)
corr_no_diag = pd.DataFrame(corr_values, index=corr_matrix.index, columns=corr_matrix.columns)
 
max_corr_val = corr_no_diag.max().max()
max_corr_pair = corr_no_diag.stack().idxmax()
 
min_corr_val = corr_no_diag.min().min()
min_corr_pair = corr_no_diag.stack().idxmin()
 
print(f"\nStrongest positive correlation: {max_corr_pair} -> {max_corr_val:.3f}")
print(f"Strongest negative correlation: {min_corr_pair} -> {min_corr_val:.3f}")
```
